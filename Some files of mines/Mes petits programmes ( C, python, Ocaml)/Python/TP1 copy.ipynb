```python
#EXERCICE4 
import numpy as np
import matplotlib.pyplot as plt 

def f(n):
    while n > 1:
        return n * f(n - 1)
    return 1

def SommeCos(x,n):
    Sn=0
    for i in range(0, n+1):
        Sn+= ((np.pow(-1,i))*(np.pow(x, 2*i)))/(f(2*i))
    return Sn


x = np.arange(0,50)
y= np.arange(0,50)
z= np.arange(0,50)
for i in range(0, 50):
        y[i]=np.pi
for i in range(0, 50):
        z[i]=SommeCos(np.pi,i)
plt.plot(x, np.cos(y), label="cos(pi)", color="#F44336", lw=2, linestyle="-")
plt.plot(x, z, label="SommeCos(pi,n)", color="blue", lw=2, linestyle="--")
plt.grid(lw=0.5)
plt.legend()
plt.xlabel("Les valeurs de n ")
plt.ylabel("Le return des fonctions")
```

```python
#EXERCICE 5
import numpy as np
import matplotlib.pyplot as plt 

def f(n):
    while n > 1:
        return n * f(n - 1)
    return 1

def Sn(n):
    Sn=0
    for i in range(1, n):
        Sn+= ((-1)**(i+1))/(i)
    return Sn
print(Sn(1))
print(Sn(2))
print(Sn(3))

x = np.arange(1, 10)
y = np.arange(0, 1)
for i in range(0, 1):
    y[i]=abs(Sn(i)-np.log(2))
plt.plot(x, y, label="sin(x)", color="#2196F3", lw=2)
```

```python
x = np.linspace(-2*np.pi, 2*np.pi, 300)

plt.figure(figsize=(10, 4))
plt.plot(x, np.sin(x), label="sin(x)", color="#2196F3", lw=2)
plt.plot(x, np.cos(x), label="cos(x)", color="#F44336", lw=2, linestyle="--")
plt.axhline(0, color="gray", lw=0.8)   # ligne y=0
plt.axvline(0, color="gray", lw=0.8)   # ligne x=0
plt.title("Fonctions Trigonométriques", fontsize=14)
plt.xlabel("x (radians)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(alpha=0.3)
plt.xticks([-2*np.pi, -np.pi, 0, np.pi, 2*np.pi],
           ["-2π", "-π", "0", "π", "2π"])
plt.yticks([-1,-0.5,0,0.5,1])
plt.tight_layout()
plt.savefig("sincos.png", dpi=150)
plt.show()
```
