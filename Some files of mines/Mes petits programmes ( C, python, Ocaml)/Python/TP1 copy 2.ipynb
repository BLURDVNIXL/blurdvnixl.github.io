```python
import numpy as np

A=np.array([[2,3],
            [2,3]])

def Stochastique(A):
    a=1
    [n,p]=np.shape(A)
    S=np.arange(0, n)
    while(a==1):
        for i in range(0, n):
            for j in range(0, p):
                S[i]+=A[i][j]
                if S[i][j]>1:
                    break
        for i in range(0, n):
            if S[i]!=1:
                break
            else :
                print("La matrice est stochastique")
                break
    a=0

Stochastique (A)      

```
