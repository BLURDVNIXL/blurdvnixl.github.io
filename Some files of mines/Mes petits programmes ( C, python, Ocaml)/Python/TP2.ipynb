```python
#EXERCICE4 
import numpy as np
import matplotlib.pyplot as plt 

def f(n):
    while n > 1:
        return n * f(n - 1)
    return 1

def SommeCos(x,n):
    Sn=0
    for i in range(0, n+1):
        Sn+= (((-1)**i)*(x**(2*i)))/(f(2*i))
        Sn+= ((np.pow(-1,i))*(np.pow(x, 2*i)))/(f(2*i))
    return Sn
```
