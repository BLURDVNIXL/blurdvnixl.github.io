```python
import numpy as np
import math as m
#Question 1
def square(n:int):
    L=[]
    for i in range(n):
        if m.pow(i,2) <= n :
           L.append(int(m.pow(i,2)))
    return L

print(square(12))
```

```python
#Question 2
def cube(n:int):
    L=[]
    for i in range(n):
        if m.pow(i,3) <= n :
           L.append(int(m.pow(i,3)))
    return L

print(cube(12))
```

```python
def div(n:int):
    D= []
    for i in range (1,n):
        if n%i ==0 :
            D.append(i)
    return D
div(10)
```

```python
def pgcd(a:int, b:int):
    if a==b:
        return a
    if a > b:
        return pgcd(b,a-b)
    if b > a:
        return pgcd(a,b-a) 
    return 0
    
print(pgcd(12,4))
```

```python
def minprem(n: int):
    L=[]
    for i in range (2,n,1):
        if pgcd(n, i) == 1 :
            L.append(i)
    return min(L)
print(minprem(10))
```

```python
def perfect(n: int):
    Div= div(n)
    k= len(div(n))
    som=0
    for i in range (0,k,1):
        som= som+ Div[i]
    if som == n:
        return True
    else:
        return False
print(True==perfect(6))
```

```python
def triangle(n: int):
    if n==3:
        return True
    for p in range(1,n):
        som=0
        for i in range(1,p):
            som+=i
        if n==som:
          return True
    else :
          return False
    
triangle(6)
```

```python
def isprime(n:int):
    Div=div(n)
    if max(Div) ==1 :
        return True
    else :
        return False    
print(isprime(2))
    
```

```python
def prime(n:int):
    L=[x for x in range(n) if isprime(x)]
    return L
prime(3)
```
