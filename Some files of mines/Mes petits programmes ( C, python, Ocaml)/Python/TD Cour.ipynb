```python
def compter_serie(terme, i):
    #Compte combien de fois terme[i] se répète consécutivement à partir de i.
    chiffre = terme[i]
    count = 0
    while i < len(terme) and terme[i] == chiffre:
        count += 1
        i += 1
    return chiffre, count

def terme_suivant(terme):
    #Génère le terme suivant à partir du terme actuel.
    resultat = ""
    i = 0
    while i < len(terme):
        chiffre, count = compter_serie(terme, i)
        resultat += str(count) + str(chiffre)
        i += count  # on saute toute la série
    return resultat

def conway(n):
    #Retourne le n-ième terme de la suite de Conway (en partant de n=1).
    terme = "1"
    for _ in range(n):
        terme = terme_suivant(terme)
    return terme
conway(0)
```

```python
n = int(input("Entrez le nombre de lignes : "))
c = int(input("Entrez le nombre de colonnes : "))
for i in range(0, n+1):
    for j in range(0, c+1):
        if j <= i:
            print(" * ", end="")
        else:
            print("   ", end="")
    print()
```
