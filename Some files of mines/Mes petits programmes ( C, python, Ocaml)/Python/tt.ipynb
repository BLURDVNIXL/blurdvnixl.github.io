```python
#pivot de Gauss simple en francais
import numpy as np
def gauss_elimination(A):
    n = A.shape[0]
    for i in range(n):
        # Trouver le pivot
        max_row_index = np.argmax(np.abs(A[i:, i])) + i
        A[[i, max_row_index]] = A[[max_row_index, i]]  # Échanger les lignes
        
        # Éliminer les éléments en dessous du pivot
        for j in range(i + 1, n):
            factor = A[j, i] / A[i, i]
            A[j] = A[j] - factor * A[i]
    return A
```

```python
def factorielle1(n):
    if n < 0:
        return None
    produit = 1
    for i in range(1, n + 1):
        produit *= i
    return produit
def factorielle2(n):
    if n < 0:
        return None
    while n > 1:
        return n * factorielle2(n - 1)
    return 1
def factorielle3(n):
    cpt = 1
    fact = 1
    if n < 0:
        return None
    while cpt <= n:
        fact *= cpt
        cpt += 1
    return fact
N=int(input("Entrez un nombre pour calculer sa factorielle: "))
print("La factorielle de", N, "est", factorielle1(N))
print("La factorielle de", N, "est", factorielle2(N))
print("La factorielle de", N, "est", factorielle3(N))
```

```python
from math import isqrt

# ─── 1. Tri personnalisé ───────────────────────────────────────────
eleves = [("Daniel", 12), ("Marie", 18), ("Kofi", 15)]

eleves.sort(key=lambda x: x[1])
print("Tri par note     :", eleves)

eleves.sort(key=lambda x: x[0])
print("Tri par nom      :", eleves)

# ─── 2. map() ─────────────────────────────────────────────────────
notes = [12, 15, 8, 18]

doubles = list(map(lambda x: x * 2, notes))
print("Notes doublées   :", doubles)

# ─── 3. filter() ──────────────────────────────────────────────────
notes = [12, 15, 8, 18, 9, 7]

admis = list(filter(lambda x: x >= 10, notes))
print("Admis (>= 10)    :", admis)

# ─── 4. sorted() sur des dicos ────────────────────────────────────
etudiants = [
    {"nom": "Daniel", "moyenne": 14},
    {"nom": "Marie",  "moyenne": 17},
    {"nom": "Kofi",   "moyenne": 11}
]

classement = sorted(etudiants, key=lambda x: x["moyenne"], reverse=True)
print("Classement       :")
for e in classement:
    print(f"   {e['nom']} → {e['moyenne']}")

# ─── 5. Fonction retournant un lambda ─────────────────────────────
def puissance(exp):
    return lambda x: x ** exp

carre = puissance(2)
cube  = puissance(3)

print("4²               :", carre(4))
print("3³               :", cube(3))
```
