```python
def auto(n):
    return all(n == n + int(str(n)[i]) for i in range(1, len(str(n)) - 1))

print((auto(20), auto(10)))
```

```python
def somme_chiffres(k):
    # Décompose k chiffre par chiffre et additionne
    s = 0
    while k > 0:
        s += k % 10
        k //= 10
    return s

def autonb(n):
    # Teste tous les entiers k de 1 à n-1
    for k in range(1, n):
        if k + somme_chiffres(k) == n:
            return False   # k est un générateur de n
    return True

(autonb(20),autonb(10))
```

```python
def ListeAuto(N):
    return [n for n in range(1, N + 1) if autonb(n)]
```

```python


def Diff(N):
    liste = ListeAuto(N)
    diffs = set()
    for i in range(len(liste) - 1):
        diffs.add(liste[i + 1] - liste[i])
    return diffs
print(Diff(10000))
```

```python
def Diff(N):
    liste = ListeAuto(N)
    diffs = []
    for i in range(len(liste) - 1):
        d = liste[i + 1] - liste[i]
        diffs.append(d) if d not in diffs else None
    return diffs
print(Diff(10000))
```

```python
def NextAuto(n):
    while not autonb(n):
        n += 1
    return n
```

```python
def Saut(N):
    liste = ListeAuto(N)
    for i in range(len(liste) - 1):
        if liste[i + 1] - liste[i] > 11:
            return liste[i]
    return 0
```

```python


def GrandSaut(N):
    liste = ListeAuto(N)
    res = []
    for i in range(len(liste) - 1):
        diff = liste[i + 1] - liste[i]
        if diff > 11:
            res.append([liste[i], liste[i + 1], diff])
    return res
```

```python
def NonAuto(N):
    res = []
    for n in range(1, N + 1):
        generateurs = []
        for k in range(1, n):
            if k + somme_chiffres(k) == n:
                generateurs.append(k)
        if len(generateurs) >= 2:
            res.append([n] + generateurs)
    return res
```

```python
def nombre2tableau(n, k):
    res = []
    for i in range(n - 1, -1, -1):
        res+=[(k >> i) & 1]
        #res+=[(k >> i) % 2]
    return res
print(nombre2tableau(2, 13))
```

```python
def ensemble_tableaux(n):
    return [nombre2tableau(n, k) for k in range(2 ** n)]

print(ensemble_tableaux(4))
```

```python
def comptage(T):
    nb0=0
    for i in range(len(T)):
        if T[i]==0:
            nb0+=1
    return (nb0, len(T) - nb0)

print(comptage([1, 0, 1, 0, 0]))
```

```python
def plageMax0(T):
    x=[]
    cpt=0
    for i in range(len(T)):
        if T[i] == 0:
            cpt+=1
        else:
            x.append(cpt)
            cpt=0
    x+=[cpt]  

    max_plage = x[0]
    for i in range(len(x)):
        if x[i]>max_plage:
            max_plage=x[i]
    return max_plage
print(plageMax0([0, 0, 1, 0, 0]))
```
