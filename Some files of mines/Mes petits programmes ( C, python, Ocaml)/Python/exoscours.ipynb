```python
def test():
    return 3,4
a= test()
a
b,c= test()
b
print("pourquoi b=3 et c=4 ?")
print("Dans le code que vous avez fourni, la fonction test() retourne un tuple contenant deux valeurs : 3 et 4.")
print("Lorsque vous appelez la fonction test(), elle renvoie ce tuple (3, 4).")
print("Ensuite, lorsque vous écrivez b, c = test(), vous utilisez une fonctionnalité de Python appelée 'unpacking' (déballage).")
print("Le déballage permet d'assigner les valeurs du tuple retourné par la fonction directement à plusieurs variables en une seule ligne.")
print("Ainsi, la première valeur du tuple (3) est assignée à la variable b, et la deuxième valeur (4) est assignée à la variable c.")
print("C'est pourquoi, après l'exécution de b, c = test(), vous obtenez b = 3 et c = 4.")
print("resume : le déballage de tuple permet d'assigner plusieurs valeurs retournées par une fonction à des variables distinctes en une seule opération.")
```

```python
# Chaque ligne = un produit, chaque colonne = quantité d'ingrédient
# Produits : [Pizza, Salade, Sandwich]
# Ingrédients : [Farine, Tomates, Fromage]
recettes = np.array([
    [200, 100, 150],  # Pizza
    [0,   200, 50],   # Salade
    [100, 50,  100]   # Sandwich
])

# On veut faire 10 pizzas, 5 salades, 8 sandwiches
commande = np.array([10, 5, 8])

# Calcul des ingrédients nécessaires (produit matriciel)
ingredients_totaux = commande @ recettes
print("Ingrédients nécessaires :")
print(f"Farine: {ingredients_totaux[0]}g")
print(f"Tomates: {ingredients_totaux[1]}g")
print(f"Fromage: {ingredients_totaux[2]}g")
```

```python
# Lignes = utilisateurs, Colonnes = films
notes = np.array([
    [5, 3, 0, 1],  # Alice
    [4, 0, 0, 1],  # Bob
    [1, 1, 0, 5],  # Claire
    [0, 0, 5, 4]   # David
])

# Trouver les utilisateurs similaires (corrélation)
from numpy.linalg import norm

def similarite(user1, user2):
    # Cosinus de similarité
    return np.dot(user1, user2) / (norm(user1) * norm(user2))

print("Similarité Alice-Bob :", similarite(notes[0], notes[1]))
```
