```python
def nombreparfait(n):
    if n < 1:
        return False
    somme = 0
    for i in range(1, n):
        if n % i == 0:
            somme += i
    return somme == n
```

```python
def somme_n_premiers(n):
    """Calcule 1 + 2 + ... + n"""
    somme = 0
    for i in range(1, n + 1):
        somme += i
    return somme


def trouver_n(a):
    """Trouve n tel que 1 + 2 + ... + n > a^5"""
    seuil = a ** 5
    somme = 0
    n = 0
    
    while somme <= seuil:
        n += 1
        somme += n
    
    return n

x=input("ENTREZ UN NOMBRE")
print(somme_n_premiers(int(x)))
print(trouver_n(int(x)))

```

```python
def liste_des_carres(n,p):
    if p < n :
        return "impossible"
    Liste=[]
    for i in range (n,p+1):
        Liste.append(i*i)
    return Liste
```
