```python
# TP 1
#QUESTION 1

import numpy as np
import matplotlib.pyplot as plt 

def f(n):
    while n > 1:
        return n * f(n - 1)
    return 1

def Sn(x,n):
    Sn=0
    for i in range(0, n+1):
        Sn+= (x**i)/f(i)
    return Sn

x= np.arange(0, 100, 1)
y=[]
for k in range(0, 100, 1):
    y.append( Sn(3,k))
coef = np.polyfit(x, y, 1)     
plt.figure(figsize=(7, 5))
plt.scatter(x, y, c=y)
plt.title("Nuage de points de Sn(3)")
plt.legend()
plt.grid(alpha=0.3)
```
