### 1-Fonction square

```python
def square(n):
    L=[]
    k=0
    while k**2<=n:
       L.append(k)
       k+=1
    return L
```

### 2-Fonction cube

```python
def cube(n):
    L=[]
    k=0
    while k**3<=n:
       L.append(k)
       k+=1
    return L
```

### 3-a-Fonction div

```python
def div(n):
    L=[]
    for i in range(1,n):
        if n%i==0:
            L.append(i)
    return L
```

### 3-b-Fonction recursive pgcd

```python
def pgcd(a,b):
    if a==b:
        return a
    elif a>b:
        return pgcd(a-b,b)
    else:
        return pgcd(a,b-a)
```

### 3-c-Fonction minprem

```python
def minprem(n):
    for m in range(2,n):
       N=[]
       M=[]
       for i in range(1,m+1):
           if m%i==0:
              M.append(i)
       for i in range(1,n+1):
           if n%i==0:
              N.append(i)  
       V=[x for x in M if x in  N]
       if len(V)==1:
        return m
```

### 3-d-Fonction perfect

```python
def perfect(n):
    L=[]
    if n==0:
        return False
    for i in range(1,n):
        if n%i==0:
            L.append(i)
    if n==sum(L):
        return True
    else:
        return False

```

### 4-Fonction triangle

```python
def triangle(n):
    if n==0:
        return False
    som=0
    i=1
    while som<n:
        som+=i
        i+=1
    if som==n:
        return True
    else:
        return False
```

### 5-a-Fonction isprime

```python
def isprime(n):
    cpt=0
    for i in range(1,n+1):
        if n%i==0:
           cpt+=1
    if cpt==2:
        return True
    else :
        return False
```

### 5-b-Fonction prime

```python
def prime(n):
    L=[]
    for i in range(1,n+1):
        if isprime(i)== True:
            L.append(i)
    return L
```

### 5-c-Fonction puissance

```python
def puissance(n):
    L=prime(n)
    for p in L:
       k=2
       while p**k<n:
         k+=1
       if p**k==n:
          return True
    return False
    
```
