```python

#1
import math as m
import numpy as np

def armstrong(n) :
    if n<0: 
        return False
    else:
        nombre= str(n)
        p= len(nombre)
        som= 0
        for i in range (0,p,1) :
            som= som+ m.pow(int(nombre[i]),p)
        if som==n:
            return True
        else:
            return False
armstrong (153)
```

```python
#PARTIE A

def fusion2(lst1,lst2):
    i1, i2= 0, 0
    n1, n2= len(lst1), len(lst2)
    lst= []#...=[]
    while i1 + i2 < n1 + n2:
        if i1 == n1:
            return lst+ lst2[0:]#...=0
        elif i2 == n2:#...=i2 ==n2
            return lst+ lst1[:n2]#...=lst1[:n2]
        elif lst1[i1] < lst2[0]:#...=lst2[0]
            lst.append(lst1[i1])
            i1+=1#...=1
        else:
            lst.append(lst2[i2])#...=lst2[i2]
            i2+=1#...=1
    return lst#...=lst

print(fusion2([4,5,6],[1,2,3]))
    
```

Nous proposons un programme qui permet de à l'utilisateur de faire une " to do list " afin de mieux s'organiser.


```python
def todolist():
    l=[]
    while True == True :
        print("\n\t\t\t\t\t\tMENU :\n\t\t\t1- Ajouter un élément\n\t\t\t2- Supprimer un élément\n\t\t\t3- Afficher liste\n\t\t\t4- Arrêter\n")
        n = int(input("Choississez une option entre 1 et 4 : "))
        while (n<1 or n>4):
            n = int(input("Option inexistante. Enrez un choix valide : "))
        if n == 1:
            a=input("Entrez l'élément à ajouter : ")
            l.append(a)
        if n == 2:
            a=input("Entrez un élément à supprimer : ")
            if a in l:
                l.remove(a)
        if n == 3:
            for i in l:
                print("\t",i)
        if n == 4:
            break
    for i in l:
       print("\t",i)
            
todolist()
```
