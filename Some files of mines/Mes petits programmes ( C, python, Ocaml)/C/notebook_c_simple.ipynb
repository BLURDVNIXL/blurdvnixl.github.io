# 📚 Collection de Programmes C Simples

## 1️⃣ Calcul de Factorielle

**Objectif** : Calculer la factorielle d'un nombre entier n

La factorielle de n (notée n!) est le produit de tous les entiers positifs de 1 à n.
- Exemple : 5! = 1 × 2 × 3 × 4 × 5 = 120

---

## 2️⃣ Suite de Fibonacci

**Objectif** : Afficher le n-ième terme de la suite de Fibonacci

La suite de Fibonacci : 0, 1, 1, 2, 3, 5, 8, 13, ...
- Chaque terme est la somme des deux termes précédents
- F(n) = F(n-1) + F(n-2)

---

## 3️⃣ Test de Primalité

**Objectif** : Vérifier si un nombre est premier ou afficher tous les nombres premiers jusqu'à n

Un nombre premier est un entier naturel supérieur à 1 qui n'a aucun diviseur positif autre que 1 et lui-même.

---

## 4️⃣ Nombre Cubique

**Objectif** : Vérifier si un nombre est cubique

Un nombre cubique est un nombre égal à la somme des cubes de ses chiffres.
- Exemple : 153 est cubique car 1³ + 5³ + 3³ = 1 + 125 + 27 = 153

---

## 5️⃣ Calcul de Moyenne

**Objectif** : Calculer la moyenne d'une série de notes saisies par l'utilisateur

On utilise une boucle infinie qui s'arrête quand on entre un nombre négatif.

---

## 6️⃣ Calcul de Racine Carrée (Méthode de Newton)

**Objectif** : Calculer la racine carrée d'un nombre avec la méthode de Newton-Raphson

Algorithme itératif : X(n+1) = (X(n) + A/X(n)) / 2

---

## 7️⃣ Échange de Variables avec Pointeurs

**Objectif** : Démontrer l'utilisation des pointeurs pour échanger deux variables

Les pointeurs permettent de modifier les valeurs originales en passant les adresses aux fonctions.
